## 10. Date of Birth Cleaning

The `date_of_birth` column is converted from text into a proper datetime format.

The dataset contains dates represented in different formats, so mixed-format parsing is used while invalid values are converted to missing values for validation.

```python
clients_df['date_of_birth'].head()
```

```python
clients_df['date_of_birth'].dtype
```

```python
clients_df['date_of_birth'] = pd.to_datetime(
    clients_df['date_of_birth'],
    format='mixed',
    dayfirst=False,
    errors='coerce'
)
```

```python
print(clients_df['date_of_birth'].dtype)
print("Invalid dates:", clients_df['date_of_birth'].isna().sum())
```

## 11. Age Feature Engineering

Age is derived from the client's date of birth using a fixed reference date.

Age is more suitable than the original date of birth for buyer segmentation because it represents a directly interpretable demographic characteristic while avoiding the use of the raw birth date as a modeling feature.

```python
reference_date = pd.Timestamp('2026-01-01')

clients_df['age'] = (
    reference_date.year - clients_df['date_of_birth'].dt.year
)
```

```python
clients_df[['date_of_birth', 'age']].head(10)
```

```python
clients_df['age'].describe()
```

```python
clients_df['age'].min(), clients_df['age'].max()
```

```python
reference_date = pd.Timestamp('2026-01-01')

clients_df['age'] = (
    (reference_date - clients_df['date_of_birth']).dt.days // 365
)
```
