# Property Dataset Exploration

## 1. Introduction

This notebook performs exploratory data analysis (EDA) on the property dataset used in the real estate buyer segmentation project.

The objective is to understand the structure, quality, data types, missing values, and basic characteristics of the property transaction data before integrating it with client information.

The property dataset contains information related to property listings and transactions, including property size, sale price, transaction date, unit category, listing status, and client references.

## 6. Missing Value Analysis

Missing values are examined to identify incomplete property records.

Missing values are especially important for the `client_ref` column because some property records may not have a recorded client reference. These records should not automatically be removed because they can still be useful for property-level analysis.

```python
property_df.isnull().sum()
```

## 7. Duplicate Record Analysis

Duplicate records are checked to determine whether the property dataset contains repeated rows.

Removing or investigating duplicates is important because duplicate transactions could distort subsequent analysis and feature calculations.

```python
property_df.duplicated().sum()
```

## 8. Client-Property Relationship

The number of unique client references and the frequency of property records associated with each client are examined.

This helps determine whether individual clients can be associated with multiple property records, which is important for later client-level feature engineering.

```python
property_df['client_ref'].nunique()
```

### Checking Rows with Missing Values

The following code displays the first 10 rows from `property_df` that contain at least one missing (`NaN`) value.
This helps us inspect the records with missing data before deciding how to handle them.

```python
property_df[property_df.isnull().any(axis=1)].head(10)
```

## 9. Key Findings

The property dataset contains 10,000 records and 9 variables.

No completely duplicated rows were identified. The `client_ref` column contains missing values, indicating that some property records do not have a recorded client association.

The available client references correspond to 2,000 unique clients. Several clients are associated with multiple property records, making it possible to aggregate property-level information into client-level behavioral and investment features.

These findings provide the foundation for integrating the property dataset with the client dataset.
